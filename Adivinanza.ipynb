```python
import random  ## traes una libreria de otro lugar automáticamente (Numeros random, aleatorio, en este caso para jugar)

def juego_adivinanza():
  numero_secreto = random.randint(1, 100) # declaración de variable, llama a la libreria y solo toma numero enteros (random.randit)
  intentos = 0 #Para que se repita el juego
  adivinado = False  ## cuando la persona cambia e true

  print("Bienvenid@ al juego de adivinanza")
  print("Tienes que adivinar un número entre 1 y 100")

  while not adivinado:  # si el bucle no esta adivinado.....
    experimento = int(input("Adivina el número: "))  ## se define la variable con una entrada de un numero(numero entero, int)
    intentos += 1 # los intentos, se incremente, para saber cuantos intentos

    if experimento < numero_secreto: # si.... experimento es menor a numero secreto...
      print("El número es mayor.")
    elif experimento > numero_secreto: ### y si el experimento es mayor al numero secreto....
      print("El numero es menor.")
    else: # demas... ademas....entonces...
      adivinado = True
      print(f" ¡Felicidades! Adivinaste el numero en {intentos} intentos.")

juego_adivinanza()
```
