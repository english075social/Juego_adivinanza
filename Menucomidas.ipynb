```python
Catálogo de comidas según el momento del día.
```

```python
import pytz
from datetime import datetime

zona = pytz.timezone("Europe/Madrid")
hora_espana = datetime.now(zona).hour

if 6 <= hora_espana < 12:
    plato = "Hora del desayuno, elige un plato: Tostada, Bollería y café o zumo"
elif 12 <= hora_espana < 15:
    plato = "Hora de la comida, elige un plato: Lentejas, croquetas y un yogurt"
elif 15 <= hora_espana < 19:
    plato = "Hora de la merienda, elige un plato: Bollicao, bocadillo o fruta"
elif 19 <= hora_espana < 23:
    plato = "Hora de la cena, elige un plato: Pasta, pizza, judías"
else:
    plato = "Es hora de dormir"

print(f" {plato}")
plato_elegido = input("Introduce el nombre del plato que prefieras")
print (f"Enhorabuena has elegido:  {plato_elegido}")
```
