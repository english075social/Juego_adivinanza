Zapatos mio explicado

```python
# Definimos una función para pedir números enteros de forma segura y evitar que el programa falle
def pedir_entero_positivo(mensaje: str) -> int:
    # Creamos un bucle infinito que no se detendrá hasta que el usuario ingrese un número válido
    while True:
        # Iniciamos un bloque 'try' para intentar ejecutar un código que podría causar un error
        try:
            # Solicitamos el texto al usuario con input() y lo convertimos a entero con int()
            valor = int(input(mensaje))
            # Verificamos si el número ingresado es estrictamente mayor que cero
            if valor > 0:
                # Si es mayor a 0, devolvemos el valor y salimos de esta función
                return valor
            # Si el número es 0 o negativo, mostramos un mensaje de advertencia
            print("❌ Por favor, ingresa un número mayor a 0.")
        # Capturamos el error si el usuario ingresó letras o texto en lugar de un número
        except ValueError:
            # Informamos al usuario sobre el tipo de dato correcto que debe ingresar
            print("❌ Entrada inválida. Debes ingresar un número entero.")


# Definimos la función principal que contendrá toda la lógica de la tienda
def gestionar_inventario():
    # Mostramos un título de bienvenida en la consola de comandos
    print("=== GESTIÓN DE INVENTARIO DE ZAPATERÍA ===")

    # Solicitamos la cantidad inicial usando nuestra función segura y la guardamos en 'inventario'
    inventario = pedir_entero_positivo("Ingrese la cantidad inicial de pares de zapatos: ")
    # Inicializamos la variable 'transacciones' en 0 para llevar el conteo de ventas
    transacciones = 0

    # Creamos un bucle que se ejecutará mientras la variable 'inventario' sea mayor a 0
    while inventario > 0:
        # Mostramos en pantalla el inventario actual disponible
        print(f"\n📦 Inventario actual: {inventario} pares de zapatos.")

        # Pedimos al usuario la cantidad de zapatos a vender de forma segura
        vendidos = pedir_entero_positivo("¿Cuántos pares de zapatos se vendieron? ")

        # Evaluamos si la cantidad a vender es mayor a la cantidad que tenemos disponible
        if vendidos > inventario:
            # Si se intenta vender más de lo existente, mostramos un mensaje de error
            print(f"⚠️ ¡Error! No puedes vender {vendidos} pares porque solo quedan {inventario}.")
            # Usamos 'continue' para saltar al inicio del bucle sin descontar inventario
            continue

        # Restamos la cantidad vendida al inventario actual (equivale a: inventario = inventario - vendidos)
        inventario -= vendidos
        # Sumamos 1 al contador de transacciones (equivale a: transacciones = transacciones + 1)
        transacciones += 1
        # Confirmamos al usuario que la venta se registró correctamente
        print(f"✅ Venta registrada con éxito. Quedan {inventario} pares.")

    # Imprimimos una línea divisoria decorativa
    print("\n-------------------------------------------")
    # Anunciamos que el inventario se ha agotado al salir del bucle 'while'
    print("🎉 ¡El inventario ha llegado a cero!")
    # Mostramos el total de ventas (transacciones) realizadas durante la ejecución
    print(f"📊 Número total de transacciones (ventas) realizadas: {transacciones}")


# Evaluamos si este script se está ejecutando directamente como programa principal
if __name__ == "__main__":
    # Llamamos a la función principal para iniciar el programa
    gestionar_inventario()
```
