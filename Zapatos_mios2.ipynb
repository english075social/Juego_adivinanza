```python
# 1. Solicitamos al usuario la cantidad inicial de zapatos y la convertimos a un número entero (int)
inventario = int(input("Ingrese la cantidad inicial de pares de zapatos en la tienda: "))

# 2. Inicializamos el contador de transacciones (ventas individuales) en 0
transacciones = 0

# 3. Iniciamos un bucle que se ejecutará de forma continua mientras el inventario sea mayor que cero
while inventario > 0:
    # 4. Solicitamos al usuario la cantidad de zapatos que se vendieron en esta transacción
    vendidos = int(input(f"¿Cuántos pares de zapatos se vendieron? (Inventario actual: {inventario}): "))

    # 5. Verificamos si la cantidad de zapatos vendidos no supera el inventario disponible
    if vendidos <= inventario:
        # 6. Restamos la cantidad vendida de nuestro inventario total
        inventario = inventario - vendidos

        # 7. Incrementamos en 1 el número de transacciones exitosas realizadas
        transacciones = transacciones + 1

        # 8. Informamos al usuario la cantidad de zapatos que quedan en el inventario
        print(f"Venta registrada. Inventario restante: {inventario} pares.")
    else:
        # 9. Si el usuario intenta vender más de lo que hay, mostramos un mensaje de advertencia
        print(f"Error: No puedes vender {vendidos} pares, solo quedan {inventario} en inventario.")

# 10. Al salir del bucle (cuando el inventario llega a 0), informamos que ya no quedan productos
print("¡El inventario se ha agotado por completo!")

# 11. Mostramos el total de transacciones de venta individuales que se realizaron
print(f"Número total de transacciones realizadas: {transacciones}")
```
