```python
import random
import time

# ==========================================
# 3. ACTUADORES / ACCIONES DEL ROBOT
# (Definición de funciones de control)
# ==========================================

def avanzar():
    print("  ➡️ [ACTUADOR] Avanzando a velocidad constante...")

def frenar():
    print("  🛑 [ACTUADOR] Frenando robot...")

def detenerse():
    print("  ⏹️ [ACTUADOR] Robot completamente detenido.")

def girar_izquierda():
    print("  ↩️ [ACTUADOR] Girando a la izquierda...")

def girar_derecha():
    print("  ↪️ [ACTUADOR] Girando a la derecha...")

def reducir_velocidad():
    print("  🐌 [ACTUADOR] Reduciendo velocidad por seguridad.")

def aumentar_velocidad():
    print("  🚀 [ACTUADOR] Aumentando velocidad.")

def seguir_ruta():
    print("  🗺️ [ACTUADOR] Siguiendo la ruta planificada...")

def buscar_ruta_alternativa():
    print("  🔄 [ACTUADOR] Recalculando y buscando ruta alternativa...")
    return True  # Simula que siempre encuentra o valida ruta alternativa

def rodear_obstaculo():
    print("  🔄 [ACTUADOR] Rodeando el obstáculo de forma segura...")

def avisar_dueno(mensaje):
    print(f"  📲 [MÓDULO COM] Notificación enviada al dueño: '{mensaje}'")

def emitir_sonido():
    print("  🔊 [ALTAVOZ] Emitiendo sonido de advertencia / interacción.")

def recoger_excrementos():
    print("  🧹 [MECANISMO] Desplegando brazo robótico para recoger excrementos...")

def guardar_bolsa():
    print("  🗑️ [MECANISMO] Guardando la bolsa de residuos en el compartimento sellado.")

def iniciar_regreso():
    print("  🏠 [NAVEGACIÓN] Iniciando ruta de regreso a casa.")

def ir_al_dock():
    print("  🔌 [NAVEGACIÓN] Acoplándose a la estación de carga (Dock)...")

def iniciar_recarga():
    print("  ⚡ [SISTEMA] Cargando batería...")

# ==========================================
# SIMULACIÓN DE SENSORES Y VARIABLES INICIALES
# ==========================================

# Variables del estado inicial
bateria = random.randint(15, 100)       # Porcentaje de batería inicial
bolsas_disponibles = random.randint(1, 5) # Cantidad de bolsas para residuos
agua_disponible = True                   # Nivel de agua suficiente
arnes_colocado = True                   # Confirmación del arnés
clima_peligroso = False                 # Estado meteorológico

# ==========================================
# 1. PREPARAR EL PASEO
# ==========================================
print("\n--- 🛠️ FASE 1: PREPARACIÓN DEL PASEO ---")
time.sleep(1)

print(f"🔋 Batería actual: {bateria}%")
if bateria < 20:
    print("⚠️ Batería baja.")
    iniciar_recarga()
else:
    print("✅ Batería suficiente para iniciar.")

print("🎒 Comprobando suministros y arnés...")
if not bolsas_disponibles > 0:
    avisar_dueno("No hay bolsas suficientes para el paseo.")
    print("🛑 SERVICIO DETENIDO.")
    exit()

if not agua_disponible:
    avisar_dueno("No hay agua suficiente.")
    print("🛑 SERVICIO DETENIDO.")
    exit()

if not arnes_colocado:
    avisar_dueno("El arnés no está colocado correctamente.")
    print("🛑 SERVICIO DETENIDO.")
    exit()

print("✅ Preparación completada con éxito.")
time.sleep(1)

# ==========================================
# 2. PLANIFICAR LA RUTA
# ==========================================
print("\n--- 🗺️ FASE 2: PLANIFICACIÓN DE LA RUTA ---")
time.sleep(1)

if clima_peligroso:
    avisar_dueno("Condiciones meteorológicas peligrosas detectadas.")
    print("❌ PASEO CANCELADO.")
    exit()

ruta_segura = random.choice([True, False])
if ruta_segura:
    print("✅ Ruta principal confirmada y segura.")
else:
    print("⚠️ Ruta principal no segura.")
    if buscar_ruta_alternativa():
        print("✅ Ruta alternativa confirmada.")
    else:
        avisar_dueno("No se encontró una ruta alternativa segura.")
        print("❌ PASEO CANCELADO.")
        exit()

duracion_paseo = 5  # Duración del ciclo de simulación (pasos de tiempo)
print(f"⏱️ Duración fija del paseo: {duracion_paseo} ciclos.")
time.sleep(1)

# ==========================================
# 3. EJECUCIÓN DEL PASEO (BUCLE PRINCIPAL)
# ==========================================
print("\n--- 🐕 FASE 3: INICIANDO PASEO AUTÓNOMO ---")
paseo_activo = True
tiempo_transcurrido = 0

while paseo_activo:
    tiempo_transcurrido += 1
    print(f"\n--- ⏳ Tiempo de paseo: Minuto {tiempo_transcurrido}/{duracion_paseo} ---")
    seguir_ruta()
    time.sleep(1)

    # 1. Controlar distancia del perro (Tensión de la correa / Distancia)
    perro_alejado = random.choice([True, False, False])  # 33% de probabilidad
    if perro_alejado:
        print("⚠️ 🐕 El perro se está alejando demasiado...")
        reducir_velocidad()
        detenerse()
        avisar_dueno("El perro no mantiene la distancia adecuada.")
        time.sleep(1)

    # 2. Detectar obstáculos (Cámara + LiDAR)
    hay_obstaculo = random.choice([True, False])
    if hay_obstaculo:
        print("🛑 🚧 ¡Obstáculo detectado con LiDAR/Cámara!")
        reducir_velocidad()
        espacio_suficiente = random.choice([True, False])
        if espacio_suficiente:
            rodear_obstaculo()
        else:
            detenerse()
            buscar_ruta_alternativa()
        time.sleep(1)

    # 3. Detectar otros animales (Cámara + IA)
    detecta_animal = random.choice([True, False, False])
    if detecta_animal:
        print("🐕‍🦺 🐱 Animal detectado en las cercanías (Cámara + IA).")
        reducir_velocidad()
        riesgo_contacto = random.choice([True, False])
        if riesgo_contacto:
            detenerse()
            print("🛑 Manteniendo distancia de seguridad con el otro animal.")
        time.sleep(1)

    # 4. Detectar necesidades fisiológicas
    hizo_necesidades = random.choice([True, False, False])
    if hizo_necesidades:
        print("💩 El perro está haciendo sus necesidades.")
        detenerse()
        recoger_excrementos()
        guardar_bolsa()
        bolsas_disponibles -= 1
        print(f"📦 Bolsas restantes: {bolsas_disponibles}")
        avanzar()
        time.sleep(1)

    # 5. Controlar batería durante el paseo
    bateria -= random.randint(3, 8)  # Consumo de batería por minuto
    print(f"🔋 Estado de la batería: {bateria}%")
    if bateria < 15:
        avisar_dueno("Batería baja (<15%). Calculando retorno.")
        if bateria >= 5:
            iniciar_regreso()
            paseo_activo = False
            break
        else:
            print("🚨 PROTOCOLO DE EMERGENCIA: Batería críticamente baja.")
            avisar_dueno("Robot detenido por emergencia de batería.")
            detenerse()
            exit()

    # 6. Control del tiempo transcurrido
    if tiempo_transcurrido >= duracion_paseo:
        print("⏰ Tiempo programado de paseo completado.")
        paseo_activo = False

    time.sleep(1)

# ==========================================
# 4. REGRESAR A CASA
# ==========================================
print("\n--- 🏠 FASE 4: REGRESO A CASA Y FINALIZACIÓN ---")
time.sleep(1)

buscar_ruta_alternativa()  # Calcular ruta de regreso
minutos_llegada = random.randint(3, 8)
avisar_dueno(f"El robot llegará aproximadamente en {minutos_llegada} minutos.")
iniciar_regreso()

time.sleep(2)  # Simular tiempo de recorrido de regreso

print("\n🏡 ¡El robot ha llegado a casa de forma segura!")
print("🏁 Paseo finalizado.")
ir_al_dock()
iniciar_recarga()
print("⚡ Proceso completado exitosamente.")
```
